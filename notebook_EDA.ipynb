# CPRI Hackathon — Exploratory Data Analysis (EDA)
This notebook reproduces the key EDA findings for physical behaviour, candidate validity rules, sensor correlations, S4 analysis, and duplicate input groups using the cleaned dataset.

```
import json
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

from src import config
from src import eda

df = eda.load_analysis_data()
print('Cleaned Training Dataset Shape:', df.shape)
print('Validity Label Distribution:
', df['Validity_Label'].value_counts())
```

## 1. Class Balance and Distribution Summary

```
val_inv_comp = eda.compare_valid_invalid(df)
print('Valid vs Invalid Summary:')
print(json.dumps(val_inv_comp['class_counts'], indent=2))
```

## 2. Sensor and Operating Feature Correlations

```
corr_matrix, corr_summary = eda.analyze_sensor_relationships(df)
print('Sensor Pair Pearson Correlations:')
print(json.dumps(corr_summary['sensor_pair_correlations'], indent=2))

plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, fmt='.4f', cmap='coolwarm')
plt.title('Feature Correlation Matrix')
plt.show()
```

## 3. Candidate Validity Rule Evaluation

```
candidate_rules = eda.evaluate_candidate_rules(df)
df_rules = pd.DataFrame(candidate_rules).T
print(df_rules)
```

## 4. Suspicious Value (Sentinel 25.0) Analysis

```
susp_analysis = eda.analyze_suspicious_values(df)
print('Sentinel 25.0 Occurrences Count:', susp_analysis['sentinel_25_count'])
print('P(Invalid | Sentinel 25.0):', susp_analysis['sentinel_25_p_invalid'])
```

## 5. Sensor S4 Detailed Analysis

```
s4_res = eda.analyze_s4(df)
print('S4 Missing Count:', s4_res['missing_count'])
print('S4 Validity Distribution of Missing:', s4_res['validity_distribution_of_missing'])
print('S4 Pearson & Spearman Correlations:')
print(pd.DataFrame(s4_res['correlations_with_other_variables']).T)
```

## 6. Duplicate Input Groups Analysis

```
dup_analysis = eda.analyze_duplicate_targets(df)
print('Total Duplicate Input Groups:', dup_analysis['total_duplicate_groups'])
print('All Groups Invalid:', dup_analysis['all_groups_invalid'])
print('Average Reference Parameter Range per Group:', dup_analysis['avg_group_ref_range'])
```
